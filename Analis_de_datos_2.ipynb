```python
df=df_raw.replace("..",np.nan)
```

```python
df.info()
```

```python
column_mapping={}
year_pattern=re.compile(r"(\d{4})\s*\[YR\d{4}\]")
for col in df.columns:
  match=year_pattern.match(col)
  if match:
    year=int(match.group(1))
    column_mapping[col]=year
  else:
    column_mapping[col]=col.strip()
df.rename(columns=column_mapping,inplace=True)
df.head()
```

```python
id_colums=["Country Name","Country Code","Series Name","Series Code"]

year_colums=[col for col in df.columns if isinstance(col,int)]

for year in year_colums:
  df[year]=pd.to_numeric(df[year],errors="coerce")
df.head()
```

```python
df.info()
```

```python
df_long=pd.melt(df,id_vars=id_colums,value_vars=year_colums,var_name="Year",value_name="Value")
df_long["Year"]=df_long["Year"].astype(int)
```

```python
df_long.sort_values(by=["Series Name","Year"],inplace=True)
df_long.head()
```

```python
df_wide=df_long.pivot_table(index="Year",
                      columns="Series Name",
                      values="Value")
df_wide.head()
```
