```python
import pandas as pd
```

```python
df_az=pd.read_csv("/content/drive/MyDrive/Data/finanzas_amazon.csv")
df_az.head()
```

```python
df_az.colums=["Date","Close","High","Low","Open","Volume"]
df_az.head()
```

```python
df_az=df_az.iloc[2:,:].reset_index(drop=True)
```

```python
df_az.head()
```

```python
df_az=df_az.rename(columns={"Price":"Date"})
```

```python
df_az.info()
```

```python
df_az["Date"]=pd.to_datetime(df_az["Date"])
df_az["Close"]=df_az.Close.astype("float")
df_az["High"]=df_az.High.astype("float")
df_az["Low"]=df_az.Low.astype("float")
df_az["Open"]=df_az.Open.astype("float")
df_az["Volume"]=df_az.Volume.astype("float")
df_az.tail(10)
```

```python
df_az[(df_az.Open<200)]
```

```python
df_az.loc[df_az.Open<100,"status"]="Atipico"
df_az
```

```python
df_az["Tendencia"]="Estable"
df_az.head()
```

```python
df_az.head()
```

```python
df_az.loc[df_az.Close>df_az.Close.shift(1),"Tendencia"]="Sube"
df_az.head(10)
```

```python
df_az.loc[df_az.Close<df_az.Close.shift(1),"Tendencia"]="Baja"
df_az.head(10)
```

```python
df_az.head()
```
